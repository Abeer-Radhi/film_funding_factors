## Technical Report

### Introduction
Film funding decisions are made before financial outcomes are known. This project examines how factors available before release, including genre, production budget, previous lead experience, runtime, and available prior creative team history, relates to a film's **reported return**.

The analysis combines IMDb film, cast, crew, and genre information with TMDB reported budget, revenue, and production-company data. The main analysis covers **2,465 selected films released from 2014 to 2023**.

### Financial Measure
**Reported Return = Reported Revenue minus Production Budget, divided by Production Budget**

This is a simplified reported-return measure. It excludes marketing, distribution, cinema/distributor shares, financing costs, and investor payments. It is therefore **not actual investor profit or a true break-even measure**.

A film is described as **below budget** when its reported revenue is lower than its reported production budget.

### Problem Statement
While film studios consider factors such as genre, budget, and previous lead experience when assessing a film's potential success before funding it, these factors showed weak relationships with reported return across **2,465 selected IMDb and TMDB films released from 2014 to 2023**. In this sample, **34.8% (858 films)** reported revenue below production budget, with an **average shortfall of $11.05 million** among those films. This creates uncertainty around which factors should receive greater consideration before funding a film.

### Objectives
1. Determine the relationship between budget, previous lead experience, genre, and reported return.
2. Identify genres with the lowest rates of films below budget.
3. Determine how the rate of films below budget varies across budget levels.
4. Compare reported return and below budget rates by previous lead experience.
5. Compare reported return across director and writer prior-history groups.
6. **Combined model:** Determine which factors show clearer relationships with reported return when considered together.

### Target Audience
**Primary:** Studio greenlight teams who decide which film projects move forward.

**Secondary:** Investors, financiers, and content producers.

### Interpretation Scope
The project is **descriptive and exploratory**. Correlations, group comparisons, and regression results show relationships within this selected historical sample; they do not establish causation, and the regression was **not tested on new films**.

### Step 2: Load the Source Files into SQLite

To work around the memory error, all five files were loaded into one SQLite database (`raw_data.db`), one table per file, reading each file in chunks so no single file is ever fully loaded into memory at once.

This only needs to run once. `raw_data.db` was used as a local working database during development. The original raw source files and the cleaned intermediate outputs are included with the submission, so rebuilding this optional database is not required to review the submitted analysis. The cell below is wrapped in `if False:` to avoid re running several gigabytes of loading by accident.


````python
# Wrapped in `if False:` - already run once, raw_data.db exists on disk
if False:
    RAW_PATH = Path("datasets/raw")
    DB_PATH = RAW_PATH / "raw_data.db"
    conn = sqlite3.connect(DB_PATH)

    # (file name, separator, table name) for each raw file
    files_to_load = [
        ("name.basics.tsv.gz", "\t", "name_basics"),
        ("title.ratings.tsv.gz", "\t", "title_ratings"),
        ("title.basics.tsv.gz", "\t", "title_basics"),
        ("title.principals.tsv.gz", "\t", "title_principals"),
        ("TMDB_movie_dataset_v11.csv", ",", "tmdb"),
    ]

    chunk_size = 500_000  # rows per chunk, small enough to avoid the MemoryError above

    for file_name, sep, table_name in files_to_load:
        print(f"Loading {file_name} into table '{table_name}'...")
        first_chunk = True
        for chunk in pd.read_csv(RAW_PATH / file_name, sep=sep, chunksize=chunk_size, low_memory=False):
            chunk.to_sql(table_name, conn, if_exists="replace" if first_chunk else "append", index=False)
            first_chunk = False
        count = pd.read_sql(f"SELECT COUNT(*) AS rows FROM {table_name}", conn)
        print(f"  -> {count['rows'][0]:,} rows loaded\n")

    conn.close()
    print("All files loaded into raw_data.db")
````

### Step 5: Review IMDb Title Columns

IMDb stores missing values as the literal text `\N`, not a true empty value, so it's replaced with a real missing value first. Any column that is missing in more than 90% of rows is dropped, since it can't meaningfully support analysis at that point.


````python
# IMDb stores missing values as the literal text '\N', not a true NaN,
# so it's replaced with a real missing value before counting anything
clean = title_basics_filtered.replace("\\N", pd.NA)
missing_pct = (clean.isnull().sum() / len(clean) * 100).round(2)
print(missing_pct)

# A column is dropped only if it's mostly empty (over 90% missing) -
# at that point it can't meaningfully support any analysis
cols_to_drop = missing_pct[missing_pct > 90].index.tolist()
print(f"\nDropped (mostly missing): {cols_to_drop}")

title_basics_clean = clean.drop(columns=cols_to_drop)

# IMDb's original column names use camelCase; renamed to snake_case here
# for consistency with TMDB's naming before the two tables are merged
title_basics_clean = title_basics_clean.rename(columns={
    "primaryTitle": "primary_title",
    "originalTitle": "original_title",
    "isAdult": "is_adult",
    "startYear": "start_year",
    "runtimeMinutes": "runtime_minutes"
})

print(f"Remaining columns: {title_basics_clean.columns.tolist()}")
````

### Step 6: Review TMDB Columns

The same missing value check was applied here. Columns unrelated to this project's objectives (genre, budget, previous lead experience, and Reported Return vs. IMDb rating) were also dropped: `homepage`, `backdrop_path`, `poster_path`, `tagline`, `overview`, `keywords`.

`title`, `original_title`, `genres`, and `runtime` exist in both tables, so the TMDB versions are prefixed with `tmdb_` to avoid confusion after merging.


````python
missing_pct = (tmdb_filtered.isnull().sum() / len(tmdb_filtered) * 100).round(2)

# not_relevant: complete columns that just don't relate to this project's
# objectives (image paths, free text descriptions have no analytical value)
not_relevant = ["homepage", "backdrop_path", "poster_path", "tagline", "overview", "keywords"]
cols_to_drop = list(set(not_relevant + missing_pct[missing_pct > 90].index.tolist()))
print(f"Dropped: {cols_to_drop}")

tmdb_clean = tmdb_filtered.drop(columns=[c for c in cols_to_drop if c in tmdb_filtered.columns])

# title, original_title, genres, and runtime exist in BOTH tables - left
# unrenamed, merging would silently overwrite one version with the other
tmdb_clean = tmdb_clean.rename(columns={
    "title": "tmdb_title",
    "original_title": "tmdb_original_title",
    "genres": "tmdb_genres",
    "runtime": "tmdb_runtime"
})

print(f"Remaining columns: {tmdb_clean.columns.tolist()}")
````

### Step 8: Merge IMDb and TMDB

IMDb `tconst` was matched to TMDB `imdb_id`. An inner join was used for the film level financial sample because the reported-return measure requires usable reported budget and revenue.

The documented cleaning funnel is:

**227,265 IMDb title records after the initial filter + 486,063 TMDB rows → 102,868 matched rows → 2,826 rows after usable reported budget/revenue and below-$1,000 exclusions → 2,465 unique IMDb films released from 2014 to 2023.**

IMDb ratings were added for descriptive context but were not required as a before funding model predictor.


````python
# how="inner" keeps only films present in BOTH tables - a film missing
# from one side can't produce a budget/revenue/rating pair anyway
merged = title_basics_clean.merge(tmdb_clean, left_on="tconst", right_on="imdb_id", how="inner")

# Ratings joined separately with how="left": a missing rating isn't a
# reason to drop a film the way a missing budget/revenue is
merged = merged.merge(title_ratings_full, on="tconst", how="left")

print(f"Merged: {merged.shape[0]:,} rows, {merged.shape[1]} columns")
print(f"Missing averageRating: {merged['averageRating'].isnull().sum():,}")
````

### Step 9: Handle Invalid Budget and Revenue Values

Rows with missing or zero budget/revenue were excluded because the reported-return calculation requires both values. Very small reported financial values also produced implausibly extreme return ratios, so the project applied a **$1,000 minimum to both reported budget and reported revenue**.

The $1,000 rule is a data quality rule for this project, not an industry definition of a valid film budget or revenue value. The excluded alternatives were not reclassified as genuine zero-revenue films.


````python
# A budget or revenue of exactly 0 in TMDB means the figure was never
# reported, not that it was genuinely zero - so these rows are excluded
invalid = ((merged["budget"] <= 0) | (merged["revenue"] <= 0)).sum()
print(f"Rows with missing/zero budget or revenue: {invalid:,}")

merged_valid = merged[(merged["budget"] > 0) & (merged["revenue"] > 0)].copy()

# Some remaining values were still unrealistically low (as low as $1-$500),
# producing extreme ROI outliers - a $1,000 minimum was agreed with the
# instructional team as a cutoff for these likely data entry errors
below_threshold = ((merged_valid["budget"] < 1000) | (merged_valid["revenue"] < 1000)).sum()
print(f"Rows below the $1,000 threshold: {below_threshold:,}")

merged_valid = merged_valid[(merged_valid["budget"] >= 1000) & (merged_valid["revenue"] >= 1000)].copy()
print(f"Remaining rows: {merged_valid.shape[0]:,}")
````

### Step 10: Remove Duplicates and Calculate Reported Return

Duplicate films were checked using the IMDb title identifier (`tconst` / final `imdb_id`). After the cleaning and date filters, the final analytical dataset contains **2,465 unique films and zero duplicate final IMDb IDs**.

The technical dataset column remains named `ROI` for compatibility with the existing code, but the presentation and interpretation use **Reported Return**:

`Reported Return = reported revenue minus production budget, divided by production budget`

A value below 0 means reported revenue is reported revenue below production budget. This is **not actual investor profit or true break-even**, because major costs and revenue sharing items are not included.


````python
# subset=["tconst"] catches the same film appearing twice even if some
# other column differs slightly between the two rows
dupes = merged_valid.duplicated(subset=["tconst"]).sum()
print(f"Duplicate rows: {dupes:,}")

merged_valid = merged_valid.drop_duplicates(subset=["tconst"], keep="first")

# Technical column name ROI stores the reported return measure:
# (reported revenue - production budget) / production budget.
# It is not actual net profit because other film costs are excluded.
# The calculation is safe here because every row has a budget of at least $1,000
merged_valid["ROI"] = (merged_valid["revenue"] - merged_valid["budget"]) / merged_valid["budget"]

print(f"Final row count: {merged_valid.shape[0]:,}")
print(merged_valid["ROI"].describe())
````

## Prior Performance Features

To support the analysis of factors available before a film funding decision, three historical performance features were created:

- **Historical Genre Reported Return:** The average Reported Return of films released in prior years within the same genre(s).
- **Historical Director Reported Return:** The average Reported Return of films released in prior years by the same director(s).
- **Historical Writer Reported Return:** The average Reported Return of films released in prior years by the same writer(s).

Only films released **before the current film's release year** were included in each calculation. Films from the same year and future years were excluded to prevent look ahead bias and ensure that each feature represents information that could have been available before the film's release.

For films associated with multiple genres, directors, or writers, the historical Reported Return was calculated separately for each associated category or person and then averaged at the film level.

Where no prior observations were available, the historical value was retained as missing rather than imputed using future information.


````python
import pandas as pd
import numpy as np

# Load the cleaned datasets
movies = pd.read_csv("datasets/cleaned/movies_cleaned_final.csv")
cast_crew = pd.read_csv("datasets/cleaned/cast_crew_filtered.csv")

# Check the size of each dataset
print("Movies shape:", movies.shape)
print("Cast/Crew shape:", cast_crew.shape)
````

### Prior Genre Reported Return

This feature measures the average Reported Return of films released in previous years within the same genre.

For films with multiple genres, historical Reported Return is calculated for each genre separately and then averaged at the film level. Films released in the same year or later are excluded.


````python
# Keep only the columns needed for genre analysis
genre_data = movies[
    ["imdb_id", "release_year", "ROI", "genres"]
].copy()

# Split films with multiple genres into separate rows
genre_data["genre"] = genre_data["genres"].fillna("").str.split(",")
genre_data = genre_data.explode("genre")

# Clean genre names
genre_data["genre"] = genre_data["genre"].str.strip()
genre_data = genre_data[genre_data["genre"] != ""]

# Calculate total ROI and number of films for each genre and year
genre_year = (
    genre_data
    .groupby(["genre", "release_year"])
    .agg(
        sum_roi=("ROI", "sum"),
        film_count=("ROI", "count")
    )
    .reset_index()
)

# Sort by genre and year before calculating historical values
genre_year = genre_year.sort_values(
    ["genre", "release_year"]
)

# Calculate cumulative ROI from previous years only
genre_year["prior_roi_sum"] = (
    genre_year.groupby("genre")["sum_roi"].cumsum()
    - genre_year["sum_roi"]
)

# Calculate the number of previous films in each genre
genre_year["prior_film_count"] = (
    genre_year.groupby("genre")["film_count"].cumsum()
    - genre_year["film_count"]
)

# Calculate historical ROI for each genre
genre_year["historical_genre_roi_component"] = (
    genre_year["prior_roi_sum"]
    / genre_year["prior_film_count"]
)

# Add the historical genre ROI back to each film
genre_data = genre_data.merge(
    genre_year[
        [
            "genre",
            "release_year",
            "historical_genre_roi_component"
        ]
    ],
    on=["genre", "release_year"],
    how="left"
)

# Average historical ROI across all genres assigned to each film
historical_genre = (
    genre_data
    .groupby("imdb_id")["historical_genre_roi_component"]
    .mean()
    .rename("historical_genre_roi")
)
````

### Prior Director and Writer Reported Return

The cast and crew dataset was linked to the movie dataset using `imdb_id`.

Directors and writers were identified using their IMDb `nconst` identifiers. Historical Reported Return was calculated using only films released before the current film's release year.

If a film had multiple directors or writers, the available historical Reported Return values were averaged at the film level.


````python
# Keep director and writer records only
crew_roles = cast_crew[
    cast_crew["category"].isin(["director", "writer"])
][
    ["imdb_id", "category", "nconst", "primaryName"]
].drop_duplicates()

# Add each film's release year and ROI
crew_movies = crew_roles.merge(
    movies[["imdb_id", "release_year", "ROI"]],
    on="imdb_id",
    how="inner"
)

# Aggregate each person's films by year
person_year = (
    crew_movies
    .groupby(["category", "nconst", "release_year"])
    .agg(
        sum_roi=("ROI", "sum"),
        film_count=("ROI", "count")
    )
    .reset_index()
)

# Sort records chronologically for each person
person_year = person_year.sort_values(
    ["category", "nconst", "release_year"]
)

# Calculate cumulative ROI from previous years only
person_year["prior_roi_sum"] = (
    person_year
    .groupby(["category", "nconst"])["sum_roi"]
    .cumsum()
    - person_year["sum_roi"]
)

# Count previous films for each person
person_year["prior_film_count"] = (
    person_year
    .groupby(["category", "nconst"])["film_count"]
    .cumsum()
    - person_year["film_count"]
)

# Calculate each person's historical ROI
person_year["historical_person_roi"] = (
    person_year["prior_roi_sum"]
    / person_year["prior_film_count"]
)

# Add historical performance back to each film person record
crew_movies = crew_movies.merge(
    person_year[
        [
            "category",
            "nconst",
            "release_year",
            "historical_person_roi"
        ]
    ],
    on=["category", "nconst", "release_year"],
    how="left"
)

# Average multiple directors or writers at the film level
historical_people = (
    crew_movies
    .groupby(["imdb_id", "category"])["historical_person_roi"]
    .mean()
    .unstack()
)

# Rename the final columns
historical_people = historical_people.rename(
    columns={
        "director": "historical_director_roi",
        "writer": "historical_writer_roi"
    }
)
````

### Merge and Validate Prior Performance Features

The historical genre, director, and writer Reported Return features were merged with the cleaned movie dataset using `imdb_id`. A sample of the resulting records was reviewed to validate the new fields, and the final analysis ready dataset was saved for use in Tableau.


````python
# Add historical genre performance
movies_analysis = movies.merge(
    historical_genre,
    on="imdb_id",
    how="left"
)

# Add historical director and writer performance
movies_analysis = movies_analysis.merge(
    historical_people,
    on="imdb_id",
    how="left"
)

# Confirm the final dataset size
print("Final dataset shape:", movies_analysis.shape)

# Review a sample of films and their historical features
display(
    movies_analysis[
        [
            "title",
            "release_year",
            "genres",
            "ROI",
            "historical_genre_roi",
            "historical_director_roi",
            "historical_writer_roi"
        ]
    ]
    .sort_values("release_year")
    .head(20)
)

# Save the final dataset without modifying the raw data
movies_analysis.to_csv(
    "datasets/cleaned/movies_analysis_final.csv",
    index=False
)

print("movies_analysis_final.csv saved successfully.")
````

**Interpretation:** Stronger prior-history groups show higher median reported return descriptively. ### Combined Model
**Method:** OLS regression considers budget, runtime, previous lead experience, genre indicators, prior-history measures, history-availability indicators, and production-company controls together. **Result:** The expanded model uses **2,442 films**, explains **9.1%** of variation in this sample (**Adjusted R² 7.5%**), and was **not tested on new films**.

**Interpretation:** Runtime shows the clearest relationship in the final robustness interpretation. Budget and the prior-history measures show limited relationships, while previous lead experience does not show a clear standalone signal. The model is exploratory, not a tool for approving or rejecting future films.


## Regression Analysis

The following models document the combined-model analysis. The technical dataset retains the column name `star_power`, but throughout the report this field is interpreted and labeled as **Previous Lead Experience**.

Post-release fields such as IMDb rating, vote count, TMDB popularity, and revenue are not used as predictors. Revenue is used only to construct the reported-return outcome.

The final model is **exploratory**: it was built and checked on the same historical sample and was not externally tested on new films.


````python
import pandas as pd
import statsmodels.api as sm
from collections import Counter
from pathlib import Path

CLEANED_PATH = Path("datasets/cleaned")

# Load the final film dataset used by both the presentation and the models.
# Starting from the saved final file makes the modeling section reproducible
# without depending on variables created in earlier cleaning cells.
df = pd.read_csv(CLEANED_PATH / "movies_analysis_final.csv")

print("Dataset shape:", df.shape)
````

### Model 1: Core Factors

This baseline model tests budget, runtime, and **Previous Lead Experience** (technical field: `star_power`) against reported return. It is retained to document the progression from a simple specification to the final combined model.


````python
# Model 1 starts with the three core numeric or binary signals used in the project.
# Rows missing one of these fields are removed only for this baseline model so
# statsmodels receives a complete design matrix.
model1_df = df[["ROI", "budget", "runtime_minutes", "star_power"]].dropna()

# A constant is added so the regression estimates an intercept rather than
# forcing the fitted relationship to pass through zero.
X1 = sm.add_constant(
    model1_df[["budget", "runtime_minutes", "star_power"]],
    has_constant="add"
)
y1 = model1_df["ROI"]

model1 = sm.OLS(y1, X1).fit()
print(model1.summary())
````

### Model 2: Genre Indicators

Genre is categorical, so it is converted to dummy variables. One genre dummy is omitted as the reference category to avoid perfect multicollinearity.


````python
# Genre is stored as comma separated text, while regression requires numeric
# predictors. Each genre is therefore converted into a binary indicator.
genre_dummies = (
    df["genres"]
    .fillna("")
    .str.get_dummies(sep=",")
)

genre_dummies.columns = [
    "genre_" + c.strip().replace(" ", "_")
    for c in genre_dummies.columns
]

# One genre indicator is omitted as the reference category. Keeping every
# indicator together with an intercept would create perfect redundancy.
if genre_dummies.shape[1] > 1:
    genre_dummies = genre_dummies.iloc[:, 1:]

X2 = pd.concat(
    [df[["budget", "runtime_minutes", "star_power"]], genre_dummies],
    axis=1
).apply(pd.to_numeric, errors="coerce")

model2_data = pd.concat([df[["ROI"]], X2], axis=1).dropna()

y2 = model2_data["ROI"]
X2_clean = sm.add_constant(model2_data.drop(columns="ROI"), has_constant="add")

model2 = sm.OLS(y2, X2_clean).fit()
print(model2.summary())
````

### Model 3: Prior Performance History

Historical genre, director, and writer return measures use **earlier release years only**. Director and writer history represents prior history available inside this selected financial dataset, **does not represent complete careers**.

Missing historical values are not treated as observed zero history without context: each value is filled with zero for modeling and paired with an availability indicator showing whether history was originally available.


````python
historical_cols = [
    "historical_genre_roi",
    "historical_director_roi",
    "historical_writer_roi"
]

model_base = df.copy()

# A missing historical ROI usually means that no earlier film history was
# available for that genre or person. Replacing it with zero keeps the film in
# the model, while a separate availability flag tells the model that the zero
# represents missing history rather than an observed historical ROI of zero.
for col in historical_cols:
    model_base[col + "_available"] = model_base[col].notna().astype(int)
    model_base[col] = model_base[col].fillna(0)

core_cols = [
    "budget",
    "runtime_minutes",
    "star_power",
    "historical_genre_roi",
    "historical_director_roi",
    "historical_writer_roi",
    "historical_genre_roi_available",
    "historical_director_roi_available",
    "historical_writer_roi_available"
]

# Combine the core factors, historical measures, availability flags, and genre
# indicators so Model 3 tests them together rather than one at a time.
X3 = pd.concat(
    [model_base[core_cols], genre_dummies],
    axis=1
).apply(pd.to_numeric, errors="coerce")

model3_data = pd.concat([model_base[["ROI"]], X3], axis=1).dropna()

y3 = model3_data["ROI"]
X3_clean = sm.add_constant(model3_data.drop(columns="ROI"), has_constant="add")

model3 = sm.OLS(y3, X3_clean).fit()
print(model3.summary())
````

### Model 4: Combined Model with Capped Reported Return

Film reported return contains unusually large values. For the final base regression, the **outcome is capped at the 1st and 99th percentile cutoffs**. The films themselves are retained.

This treatment is used for the regression only. Descriptive group charts use medians of the observed current-film reported-return values unless otherwise stated.


````python
# ROI contains a small number of extreme values. For the final regression,
# only the target is capped at the 1st and 99th percentiles. This reduces the
# influence of extreme returns without changing the descriptive analyses,
# which continue to use the observed ROI values.
roi_low = model_base["ROI"].quantile(0.01)
roi_high = model_base["ROI"].quantile(0.99)
model_base["ROI_winsorized"] = model_base["ROI"].clip(roi_low, roi_high)

# Model 4 keeps the same predictors as Model 3. The main change is the capped
# ROI target, allowing a fair comparison with the later company control model.
X4 = pd.concat(
    [model_base[core_cols], genre_dummies],
    axis=1
).apply(pd.to_numeric, errors="coerce")

model4_data = pd.concat(
    [model_base[["ROI_winsorized"]], X4],
    axis=1
).dropna()

y4 = model4_data["ROI_winsorized"]
X4_clean = sm.add_constant(
    model4_data.drop(columns="ROI_winsorized"),
    has_constant="add"
)

model4 = sm.OLS(y4, X4_clean).fit()

print("N:", int(model4.nobs))
print("R²:", round(model4.rsquared, 4))
print("Adjusted R²:", round(model4.rsquared_adj, 4))
print("Overall p value:", model4.f_pvalue)
print(model4.summary())
````

**Model 4 comparison result:** N = **2,442**; R² = **0.0470**; Adjusted R² = **0.0352**. Runtime shows the clearest relationship in this specification; the other main factors show limited independent relationships.


### Model 5: Production Company Context

The expanded model used in the presentation adds production-company context to Model 4. Production companies are known before release, but they are treated as **context/control variables**, not approval rules.

Only companies appearing in at least 50 films are included to limit sparse indicators. Films may list more than one production company, so these controls can overlap.


````python
# A film can list several production companies in one text field. Split the
# field into a list so each company can be tested as its own binary control.
model_base["company_list"] = (
    model_base["production_companies"]
    .fillna("")
    .apply(lambda x: [c.strip() for c in x.split(",") if c.strip()])
)

# Count each company once per film. Using set(companies) prevents accidental
# duplicate company names within one film from increasing the count.
company_counts = Counter(
    company
    for companies in model_base["company_list"]
    for company in set(companies)
)

# Keep companies appearing in at least 50 films. This limits sparse controls
# that would add many parameters with very little supporting data.
major_companies = sorted([
    company
    for company, count in company_counts.items()
    if count >= 50
])

company_dummies = pd.DataFrame(
    {
        "company_" + company: model_base["company_list"].apply(
            lambda companies: int(company in companies)
        )
        for company in major_companies
    },
    index=model_base.index
)

print("Production company controls:", len(major_companies))
print(major_companies)
````

````python
# Add production company controls to the exact predictor set used in Model 4.
# No original factor is removed, so the change in model fit can be attributed
# to the added production company context.
X5 = pd.concat(
    [X4, company_dummies],
    axis=1
).apply(pd.to_numeric, errors="coerce")

# Use exactly the same film rows as Model 4. Holding the sample constant makes
# the R squared comparison fair and avoids improvement caused by sample changes.
model5_data = pd.concat(
    [model_base[["ROI_winsorized"]], X5],
    axis=1
).loc[model4_data.index].dropna()

y5 = model5_data["ROI_winsorized"]
X5_clean = sm.add_constant(
    model5_data.drop(columns="ROI_winsorized"),
    has_constant="add"
)

model5 = sm.OLS(y5, X5_clean).fit()

print("BASE MODEL")
print("N:", int(model4.nobs))
print("R²:", round(model4.rsquared, 4))
print("Adjusted R²:", round(model4.rsquared_adj, 4))

print("\nWITH PRODUCTION COMPANY CONTROL")
print("N:", int(model5.nobs))
print("R²:", round(model5.rsquared, 4))
print("Adjusted R²:", round(model5.rsquared_adj, 4))
print("Overall p value:", model5.f_pvalue)

print("\nCHANGE")
print("R² change:", round(model5.rsquared - model4.rsquared, 4))
print("Adjusted R² change:", round(model5.rsquared_adj - model4.rsquared_adj, 4))
````

**Expanded model result used in the presentation:** N = **2,442**; R² = **0.0906 (9.1%)**; Adjusted R² = **0.0751 (7.5%)**.

The model explains only a limited share of differences in reported return within this historical sample and was **not tested on new films**. It should therefore be interpreted as exploratory rather than predictive.


### Robustness Check

Ordinary OLS standard errors can be sensitive to unequal error variance, which is plausible in film-return data. The final interpretation therefore uses a **heteroskedasticity-robust (HC3) error check** in addition to the original OLS specification.

The purpose of this check is not to turn p-values into a greenlight rule. It is to avoid presenting a factor as a clear standalone signal when its inference changes under a reasonable alternative error specification.


````python
# HC3 robustness check for the same expanded Model 5 specification.
# Coefficients and model fit are unchanged; HC3 changes the estimated
# uncertainty around the coefficients.

model5_hc3 = model5.get_robustcov_results(cov_type="HC3")

hc3_params = pd.Series(model5_hc3.params, index=model5.params.index)
hc3_pvalues = pd.Series(model5_hc3.pvalues, index=model5.params.index)

main_factors = [
    "budget",
    "runtime_minutes",
    "star_power",
    "historical_genre_roi",
    "historical_director_roi",
    "historical_writer_roi"
]

factor_results = pd.DataFrame({
    "Coefficient": model5.params[main_factors],
    "OLS_P_Value": model5.pvalues[main_factors],
    "HC3_P_Value": hc3_pvalues[main_factors]
})

factor_results["Presentation_Label"] = [
    "Limited relationship",
    "Clear relationship",
    "No clear standalone signal",
    "Limited relationship",
    "Limited relationship",
    "Limited relationship"
]

print(factor_results.round(4))
````

Using the stronger HC3 error check, **runtime** remains the clearest factor. **Previous Lead Experience** does not provide a clear standalone signal under the robustness check, even though the ordinary OLS p-value was smaller in one specification. Budget and the historical genre, director, and writer measures remain limited.

This is the interpretation used in the final presentation. The model does not establish causation and was not tested on new films.

## Statistical Checks

Formal tests are kept in the technical report as supporting checks rather than as the main presentation story.

Because reported return is strongly skewed and contains unusually large values:
- descriptive group comparisons emphasize **median reported return**;
- the combined regression caps the **outcome** at the 1st and 99th percentile cutoffs while keeping films;
- director/writer prior-history sensitivity is recalculated using capped earlier-film return values;
- the expanded regression is also checked with **HC3 robust standard errors**.

A p-value does not prove practical importance, causation, or predictive usefulness.


````python
from scipy import stats
from pathlib import Path
import pandas as pd

# Load the final analysis file if this section is run independently.
if "df" not in globals():
    cleaned_candidates = [
        Path("datasets/cleaned/movies_analysis_final.csv"),
        Path("datasets/cleaned/movies_analysis_final(7).csv"),
        Path("movies_analysis_final.csv"),
        Path("movies_analysis_final(7).csv")
    ]
    analysis_file = next((p for p in cleaned_candidates if p.exists()), None)
    if analysis_file is None:
        raise FileNotFoundError(
            "Final analysis CSV not found. Place movies_analysis_final.csv in datasets/cleaned/."
        )
    df = pd.read_csv(analysis_file)

significance_results = []

# Objective 2: Genre
genre_cols = [c for c in df.columns if c.startswith("genre_")]
genre_groups = [
    df.loc[df[c] == 1, "ROI"].dropna()
    for c in genre_cols
    if (df[c] == 1).sum() >= 25
]

h_genre, p_genre = stats.kruskal(*genre_groups)
significance_results.append(
    ("Genre (Objective 2)", "Kruskal-Wallis", h_genre, p_genre)
)

# Objective 3: Budget level
bins = [0, 5e6, 10e6, 25e6, 50e6, 100e6, float("inf")]
tier_labels = [
    "Under $5M",
    "$5M to $10M",
    "$10M to $25M",
    "$25M to $50M",
    "$50M to $100M",
    "$100M+"
]

budget_tier = pd.cut(
    df["budget"],
    bins=bins,
    labels=tier_labels,
    right=False
)

budget_groups = [
    df.loc[budget_tier == label, "ROI"].dropna()
    for label in tier_labels
]

h_budget, p_budget = stats.kruskal(*budget_groups)
significance_results.append(
    ("Budget level (Objective 3)", "Kruskal-Wallis", h_budget, p_budget)
)

# Objective 4: Previous lead experience
lead_experience = df.loc[df["star_power"] == 1, "ROI"].dropna()
no_lead_experience = df.loc[df["star_power"] == 0, "ROI"].dropna()

u_lead, p_lead = stats.mannwhitneyu(
    lead_experience,
    no_lead_experience,
    alternative="two-sided"
)

significance_results.append(
    (
        "Previous lead experience (Objective 4)",
        "Mann-Whitney U",
        u_lead,
        p_lead
    )
)

# Objective 5: Director and writer prior history
for col, role in [
    ("historical_director_roi", "Director"),
    ("historical_writer_roi", "Writer")
]:
    role_df = df[df[col].notna()]
    role_median = role_df[col].median()

    higher = role_df.loc[role_df[col] >= role_median, "ROI"].dropna()
    lower = role_df.loc[role_df[col] < role_median, "ROI"].dropna()

    u_role, p_role = stats.mannwhitneyu(
        higher,
        lower,
        alternative="two-sided"
    )

    significance_results.append(
        (
            f"{role} prior history (Objective 5)",
            "Mann-Whitney U",
            u_role,
            p_role
        )
    )

significance_table = pd.DataFrame(
    significance_results,
    columns=["Comparison", "Test", "Statistic", "P_Value"]
)

significance_table["Significant (p < 0.05)"] = (
    significance_table["P_Value"] < 0.05
)

print(significance_table.round(4).to_string(index=False))
````

**Interpretation:** Formal group tests are secondary to the descriptive findings and should not be read as proof that a factor should be used for approval decisions.

Previous lead experience is especially important to interpret cautiously: the descriptive median reported return differs between the two groups, but below budget rates are nearly identical, the standalone group test is not statistically significant, and the stronger regression error check does not support a clear standalone signal.

The director/writer history pattern is also treated cautiously because the history fields cover only part of the sample and are not full career histories.


## Final Verification

This section is designed to run directly from the submitted final CSV files. It verifies the key sample facts requested in the mock feedback and checks whether the creative team prior-history conclusion changes when unusually large **earlier-film** reported returns are capped at the 1st and 99th percentile cutoffs.

The cap is applied when constructing the historical director/writer averages, not to the current-film median reported return shown in the presentation.


````python
from pathlib import Path
import pandas as pd
import numpy as np

# Locate submitted files either beside the notebook or in datasets/cleaned.
candidates = [Path("."), Path("datasets/cleaned"), Path("/mnt/data")]
base = next(
    p for p in candidates
    if (p / "movies_analysis_final(7).csv").exists()
    or (p / "movies_analysis_final.csv").exists()
)

movie_file = (
    base / "movies_analysis_final(7).csv"
    if (base / "movies_analysis_final(7).csv").exists()
    else base / "movies_analysis_final.csv"
)
crew_file = (
    base / "cast_crew_filtered(7).csv"
    if (base / "cast_crew_filtered(7).csv").exists()
    else base / "cast_crew_filtered.csv"
)

movies_check = pd.read_csv(movie_file)
crew_check = pd.read_csv(crew_file)

print("FINAL SAMPLE CHECK")
print("Films:", len(movies_check))
print("Years:", int(movies_check["release_year"].min()), "-", int(movies_check["release_year"].max()))
print("Duplicate imdb_id:", int(movies_check["imdb_id"].duplicated().sum()))
print("Missing runtime:", int(movies_check["runtime_minutes"].isna().sum()))
print("Below-budget films:", int((movies_check["ROI"] < 0).sum()))
print("Below-budget rate:", round((movies_check["ROI"] < 0).mean() * 100, 1), "%")
shortfall = (movies_check.loc[movies_check["ROI"] < 0, "budget"]
             - movies_check.loc[movies_check["ROI"] < 0, "revenue"]).mean()
print("Average shortfall among below-budget films: $", round(shortfall / 1e6, 2), "M")
print("Director history coverage:", int(movies_check["historical_director_roi"].notna().sum()))
print("Writer history coverage:", int(movies_check["historical_writer_roi"].notna().sum()))

# Cap earlier-film ROI at the same 1st/99th percentile cutoffs used as the
# regression sensitivity treatment, then rebuild director/writer history.
lo, hi = movies_check["ROI"].quantile([0.01, 0.99])
movies_check["ROI_capped_for_history"] = movies_check["ROI"].clip(lo, hi)

print("\nHISTORY SENSITIVITY CUT-OFFS")
print("1st percentile:", round(lo, 4))
print("99th percentile:", round(hi, 4))

def rebuild_capped_history(role):
    credits = (
        crew_check.loc[crew_check["category"].eq(role), ["imdb_id", "nconst"]]
        .drop_duplicates()
    )
    x = credits.merge(
        movies_check[["imdb_id", "release_year", "ROI_capped_for_history"]],
        on="imdb_id",
        how="inner"
    )

    person_year = (
        x.groupby(["nconst", "release_year"], as_index=False)
        .agg(
            year_roi_sum=("ROI_capped_for_history", "sum"),
            year_n=("ROI_capped_for_history", "size")
        )
        .sort_values(["nconst", "release_year"])
    )

    person_year["prior_sum"] = (
        person_year.groupby("nconst")["year_roi_sum"].cumsum()
        - person_year["year_roi_sum"]
    )
    person_year["prior_n"] = (
        person_year.groupby("nconst")["year_n"].cumsum()
        - person_year["year_n"]
    )
    person_year["history_capped"] = (
        person_year["prior_sum"] / person_year["prior_n"].replace(0, np.nan)
    )

    film_history = (
        x[["imdb_id", "nconst", "release_year"]]
        .drop_duplicates()
        .merge(
            person_year[["nconst", "release_year", "history_capped"]],
            on=["nconst", "release_year"],
            how="left"
        )
        .groupby("imdb_id", as_index=False)["history_capped"]
        .mean()
    )

    result = (
        movies_check[["imdb_id", "ROI"]]
        .merge(film_history, on="imdb_id", how="left")
        .dropna(subset=["history_capped"])
    )

    split = result["history_capped"].median()
    result["history_group"] = np.where(
        result["history_capped"] >= split,
        "Stronger history",
        "Weaker history"
    )

    summary = (
        result.groupby("history_group")
        .agg(
            films=("ROI", "size"),
            median_current_reported_return=("ROI", "median"),
            below_budget_rate=("ROI", lambda s: (s < 0).mean() * 100)
        )
        .round(3)
    )
    return result, summary

for role in ["director", "writer"]:
    role_data, role_summary = rebuild_capped_history(role)
    print(f"\n{role.upper()} HISTORY AFTER CAPPING EARLIER-FILM RETURNS")
    print("Coverage:", len(role_data))
    print(role_summary)
````

### Verification Result

The sensitivity check leaves the presentation's creative team pattern unchanged:

- **Directors:** 633 films; stronger-history group **317 films, median reported return 1.52** vs weaker-history group **316 films, 0.67**.
- **Writers:** 792 films; stronger-history group **396 films, median reported return 1.91** vs weaker-history group **396 films, 0.58**.

This supports keeping the descriptive Creative Team slide, with the stated limitation that prior history is available only for part of the sample and is not full career history.

## Limitations

### Primary Limitations

1. **Simplified Reported Return Measure**  
   Reported return compares reported revenue with production budget only. It excludes marketing, distribution, cinema/distributor shares, financing costs, and investor payments, so it is **not actual investor profit or true break-even**.

2. **Selected Financial Sample**  
   The analysis contains selected 2014 to 2023 films with usable reported budget and revenue data, not all films.

3. **Extreme Return Values**  
   Unusually large reported-return values can affect averages and rankings. Descriptive charts therefore emphasize medians; the regression caps the outcome at the 1st and 99th percentile cutoffs while keeping films.

4. **Limited Creative-Team History**  
   Director history is available for 633 films and writer history for 792 films. These fields use earlier year films available in this selected dataset and **do not represent complete careers**.

5. **Model Not Externally Tested**  
   The final model was built and checked on the same **2,442 films** and explains a limited share of variation. It was not validated on unseen films and is **not a prediction or approval tool**.

### Secondary Technical Notes
- Films can belong to multiple genres, so genre groups overlap.
- Films can list multiple production companies, so company indicators overlap.
- 23 films have missing runtime and are excluded from runtime-specific comparisons/model rows requiring runtime.
- Budget and revenue are treated as nominal reported USD values and were not adjusted for inflation.
- Budget/revenue values below $1,000 were excluded as a project data quality rule.
- Same-year releases are excluded from prior-history calculations.
- Correlations, group comparisons, and regression coefficients describe relationships and do not establish causation.


## Assumptions

1. Reported budget and revenue values are treated as comparable reported financial figures for this exploratory analysis.
2. The reported return measure compares reported revenue with production budget only and does not represent actual investor profit or break even.
3. The $1,000 minimum for reported budget and revenue is a project data quality rule used to exclude implausible or unusable financial entries.
4. Films may belong to more than one genre, so genre groups are not mutually exclusive.
5. Previous lead experience indicates whether the selected lead had earlier qualifying credits in the available data. It is not a measure of celebrity or star power.
6. Director and writer prior history uses earlier year films available in the selected financial dataset. Same year releases are excluded, and the history does not represent complete careers.
7. The regression is exploratory. It was built and checked on the same 2,442 films and was not tested on new films.
